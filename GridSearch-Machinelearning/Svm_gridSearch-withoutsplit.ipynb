```python
import pandas as pd
dataset= pd.read_csv("insurance_pre.csv")

```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset,drop_first=True).astype(int)
```

```python
independent=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dependent=dataset['charges']
```

```python

from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
svrparams = {'kernel':['linear', 'poly', 'rbf', 'sigmoid'],'gamma':['scale', 'auto'],'C':[10,100,1000,2000,3000]}
grid = GridSearchCV(SVR(),svrparams,refit=True,verbose = 3,n_jobs=-1)
grid.fit(independent,dependent)

```

```python
gridresults = grid.cv_results_
table = pd.DataFrame.from_dict(gridresults)
```

```python
age_input=float(input("Age:"))
bmi_input=float(input("BMI:"))
children_input=float(input("Children:"))
sex_male_input=int(input("Sex Male 0 or 1:"))
smoker_yes_input=int(input("Smoker Yes 0 or 1:"))
```

```python
Prediction=grid.predict([[age_input,bmi_input,children_input,sex_male_input,smoker_yes_input]])
```
