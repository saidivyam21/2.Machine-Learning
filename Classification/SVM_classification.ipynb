```python
import pandas as pd
dataset = pd.read_csv("Social_Network_Ads.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset,drop_first = True).astype(int)
```

```python
dataset.columns
```

```python
dataset=dataset.drop("User ID",axis=1)
```

```python
Independent=dataset[[ 'Age', 'EstimatedSalary', 'Gender_Male']]
dependent = dataset[[ 'Purchased']]
```

```python
dataset[[ 'Purchased']].value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(Independent,dependent,test_size=0.30,random_state = 0)

'''from sklearn.svm import SVC
classifier = SVC(C=1.0, kernel='rbf', degree=3, gamma='scale', coef0=0.0, )
classifier.fit(x_train,y_train)'''
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
svcparams =  {'kernel':[ 'poly', 'rbf', 'sigmoid'],'gamma':['scale', 'auto'],'C':[10,100,1000,2000,3000]}
```

```python
svcgrid = GridSearchCV(SVC(),svcparams ,refit=True,verbose = 3,n_jobs=-1)
svcgrid.fit(x_train,y_train)
```

```python
y_predict = svcgrid.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_predict)
```

```python
cm
```

```python
from sklearn.metrics import classification_report
```

```python
cr = classification_report(y_test,y_predict)
```
